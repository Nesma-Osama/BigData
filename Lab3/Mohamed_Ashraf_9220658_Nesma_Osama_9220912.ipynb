```python
from pyspark.sql import SparkSession
from pyspark.conf import SparkConf
```

```python
spark=SparkSession.builder.master("local[*]").appName("vodafone_app").getOrCreate()
sc=spark.sparkContext
```

```python
vodafone_rdd=sc.textFile("./in.csv")
```

```python
vodafone_rdd.take(10)
```

```python
vodafone_split_rdd=vodafone_rdd.map(lambda x: x.split(","))
```

```python
vodafone_split_rdd.take(10)
```

```python
vodafone_keys_rdd=vodafone_split_rdd.map(lambda x:(x[0].strip(),int(x[1].strip())))
```

```python
vodafone_keys_rdd.take(10)
```

```python
result_rdd=vodafone_keys_rdd.reduceByKey(lambda x,y: x+y).sortByKey(ascending=False)
```

```python
output_rdd=result_rdd.map(lambda x: f"{x[0]},{x[1]}")
```

```python
output_rdd.coalesce(1).saveAsTextFile("output")
```
